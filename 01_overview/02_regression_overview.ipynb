여러 개의 feature를 가지고 연속된 target 하나를 예측하는 것을 회귀(regression) 예측이라고 한다.

## 최근접 이웃 회귀 모델

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
sklearn.__version__
```

```python
plt.rc('font',family='Malgun Gothic')
plt.rc('axes',unicode_minus=True)
```

```python
# 농어 길이가 입력(특성)이고 무게가 라벨 타겟이 된다.
perch_length = np.array([
    8.4, 13.7, 15.0, 16.2, 17.4, 18.0, 18.7, 19.0, 19.6, 20.0,
    21.0, 21.0, 21.0, 21.3, 22.0, 22.0, 22.0, 22.0, 22.0, 22.5,
    22.5, 22.7, 23.0, 23.5, 24.0, 24.0, 24.6, 25.0, 25.6, 26.5,
    27.3, 27.5, 27.5, 27.5, 28.0, 28.7, 30.0, 32.8, 34.5, 35.0,
    36.5, 36.0, 37.0, 37.0, 39.0, 39.0, 39.0, 40.0, 40.0, 40.0,
    40.0, 42.0, 43.0, 43.0, 43.5, 44.0])
perch_weight = np.array([
    5.9, 32.0, 40.0, 51.5, 70.0, 100.0, 78.0, 80.0, 85.0, 85.0,
    110.0, 115.0, 125.0, 130.0, 120.0, 120.0, 130.0, 135.0, 110.0,
    130.0, 150.0, 145.0, 150.0, 170.0, 225.0, 145.0, 188.0, 180.0,
    197.0, 218.0, 300.0, 260.0, 265.0, 250.0, 250.0, 300.0, 320.0,
    514.0, 556.0, 840.0, 685.0, 700.0, 700.0, 690.0, 900.0, 650.0,
    820.0, 850.0, 900.0, 1015.0, 820.0, 1100.0, 1000.0, 1100.0,
    1000.0, 1000.0])
print(perch_length.shape,perch_weight.shape)
```

```python
# 두 연속형 데이터를 산포도로 시각화
plt.scatter(perch_length,perch_weight)
plt.xlabel('length')
plt.ylabel('weight')
plt.show()
```

```python
from sklearn.model_selection import train_test_split

# 입력 데이터를 2차원으로 변경
perch_length = perch_length.reshape(-1,1)
perch_length.shape

x_train,x_test,y_train,y_test = train_test_split(
    perch_length,
    perch_weight,       
    random_state=0
)

# stratify : 타겟 데이터의 비율에 맞춰 스플릿
print(x_train.shape,y_train.shape)
print(x_test.shape,y_test.shape)
```

```python
# 모델 학습
from sklearn.neighbors import KNeighborsRegressor

# 주어진 데이터와 가장 가까운 이웃 n개를 찾고, 이웃의 평균 값 예측
kn_reg = KNeighborsRegressor()
kn_reg.fit(x_train,y_train)
```

```python
# 회귀 평가(결정계수: 1에 가까울수록 좋은 예측)
print(kn_reg.score(x_train,y_train))
print(kn_reg.score(x_test,y_test))
```

```python
# 모델 예측
y_pred = kn_reg.predict(x_test[:1])
y_pred
```

```python
# 이웃 조회
_, index = kn_reg.kneighbors(x_test[:1])
index
```

```python
# 시각화
plt.scatter(x_train,y_train)
plt.scatter(x_train[index[0]],y_train[index[0]],label='neighbor')
plt.scatter(x_test[0],y_test[0],label='y_true')
plt.scatter(x_test[0],y_pred,label='y_pred')

plt.xlabel('length')
plt.ylabel('weight')
plt.legend()
plt.show()
```

## 최근접이웃모델의 한계
- 범위를 벗어난 예측이 불가능하다
- 데이터의 전체적인 경향을 학습하지 못한다.
- 국소적인 정보에만 의존한다.

```python
kn_reg = KNeighborsRegressor()
kn_reg.fit(x_train,y_train)
y_pred = kn_reg.predict([[50],[70],[300]])
y_pred
```

```python
# 이웃 조회
x = 300
_, index = kn_reg.kneighbors([[x]])
y_pred = kn_reg.predict([[x]])
```

```python
# 시각화
plt.scatter(x_train,y_train)
plt.scatter(x_train[index[0]],y_train[index[0]],label='neighbor')
plt.scatter(x,y_pred,label='y_pred')

plt.xlabel('length')
plt.ylabel('weight')
plt.legend()
plt.show()
```

## 선형 회귀

```python
from sklearn.linear_model import LinearRegression

# 모델 학습
lin_reg = LinearRegression()
lin_reg.fit(x_train,y_train)        # ax + b에 대한 값을 구함
```

```python
# 예측
y_pred = lin_reg.predict([[50],[70],[100]])
y_pred
```

```python
# 시각화
plt.scatter(x_train,y_train)
plt.scatter([[50],[70],[100]],y_pred,label='y_pred')

plt.plot([[10],[100]],[lin_reg.predict([[10]]),lin_reg.predict([[100]])])

plt.xlabel('length')
plt.ylabel('weight')
plt.legend()
plt.show()
```

```python
# 학습된 회귀 계수와 절편
# 회귀 계수는 특성마다 만들어짐 
# y = wx + b (w1*x + w2*x + ....)
print('w: ',lin_reg.coef_)
print('b: ',lin_reg.intercept_)

pred = lin_reg.coef_*[[50],[70],[100]] + lin_reg.intercept_
pred
```

# 3개의 feature로 무게 예측

```python
perch_df = pd.read_csv('data/perch_full.csv')
perch_df.head()
```

```python
# 입력/ 라벨 분리
x = perch_df[['length' ,'height','width']].to_numpy()
y = perch_df['weight'].to_numpy()

# 학습/테스트 분리
x_train,x_test,y_train,y_test = train_test_split(
    x,
    y,
    random_state=0
)

print(x_train.shape, y_train.shape)
print(x_test.shape, y_test.shape)
```

```python
# 모델 학습
lin_reg = LinearRegression()
lin_reg.fit(x_train,y_train)
```

```python
# 모델 평가
print(lin_reg.score(x_train,y_train))
print(lin_reg.score(x_test,y_test))
```

```python
# 회귀 계수, 절편 확인
print(lin_reg.coef_,lin_reg.intercept_)

x = x_test[0]
w = lin_reg.coef_
b = lin_reg.intercept_

print((w[0]*x[0] + w[1]*x[1] + w[2]*x[2])+b)
print(w @ x + b)
print(lin_reg.predict(x_test[:1]))
```

# 특성 공학 (Feature Engineering)
- 특성공학이란? 모델이 데이터를 더 잘 이해할 수 있도록 입력 데이터를 가공하고 변환하는 과정이다. 예시: 길이만 사용하는 것보다 길이²나 길이×높이를 추가하면 패턴을 더 잘 학습할 수 있다.

1. 특성 생성
- 새로운 변수 도출
기존 데이터를 조합하여 의미 있는 새로운 변수를 만든다.
예시: 키와 몸무게로 BMI 생성

- 다항특성생성
기존 변수의 제곱, 세제곱 등을 추가하여 비선형 관계를 표현한다.
예시: 길이 → 길이² 추가

- 상호작용특성
두 변수의 곱을 통해 변수 간 관계를 반영한다.
예시: 길이 × 높이 → 부피와 유사한 의미 생성

2. 특성 변환
- 스케일링
변수들의 범위를 맞춰 모델이 공정하게 학습하도록 한다.
예시: 길이(10~100)와 무게(100~1000)를 0~1 범위로 변환

- 로그변환
값의 분포가 한쪽으로 치우친 경우 이를 완화한다.
예시: 매출 데이터에 log 적용하여 큰 값 영향 줄이기

- 이산화
연속형 데이터를 구간으로 나누어 범주형으로 변환한다.
예시: 나이 → 10대, 20대, 30대

3. 차원 축소
- 주성분 분석
여러 변수를 몇 개의 핵심 변수로 압축한다.
예시: 10개 변수 → 2~3개로 축소

- 선택적 특성 제거
중요하지 않은 변수를 제거하여 모델을 단순화한다.
예시: 영향 없는 컬럼 제거

- 결측치 처리
결측값 대체
비어있는 값을 적절한 값으로 채운다.
예시: 나이 결측값 → 평균값으로 대체

4. 범주형 변수 처리
- 더미인코딩
범주형 데이터를 0과 1로 변환한다.
예시: 성별 → 남자(1,0), 여자(0,1)

- 라벨인코딩
범주를 숫자로 매핑한다.
예시: 등급 → A=0, B=1, C=2

6. 도메인 지식 활용 문제 영역에 대한 지식을 활용하여 의미 있는 특성을 만든다.
예시: 생선 데이터에서 길이, 높이, 너비를 조합해 부피 개념 생성

# 다항 처리

```python
# 다항식 추가 클래스
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=2)     # 기본 값 degree = 2
x=[[2,3]]
x_poly = poly.fit_transform(x)
print(x_poly)
print(poly.get_feature_names_out())
# bias 값 1,기존 특성 값 2개, 서로 곱한 값 1개, 각각 제곱한 값 2개 -> 6개
```

```python
poly= PolynomialFeatures(degree=3,include_bias=False)
X = [[2,3],[4,5]]
x_poly = poly.fit_transform(X)
print(x_poly)
print(poly.get_feature_names_out())
```

```python
# 농어 데이터에 2차 다항식 적용
poly = PolynomialFeatures(degree=2,include_bias=False)
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)

# 3개의 특성을 2차 다항식을 적용하여 9개의 특성으로 변경
print(x_train.shape,x_train_poly.shape)
print(x_test.shape,x_test_poly.shape)

pd.DataFrame(x_train_poly,columns=poly.get_feature_names_out()).head()
```

```python
# 모델 학습
lin_reg = LinearRegression()
lin_reg.fit(x_train_poly,y_train)

# 학습된 회귀 계수, 절편
print(lin_reg.coef_)
print(lin_reg.intercept_)
```

```python
# 모델 평가
print(lin_reg.score(x_train_poly,y_train))
print(lin_reg.score(x_test_poly,y_test))
```

```python
# 농어 데이터에 3차 다항식 적용
poly = PolynomialFeatures(degree=3,include_bias=False)
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)

# 3개의 특성을 2차 다항식을 적용하여 9개의 특성으로 변경
print(x_train.shape,x_train_poly.shape)
print(x_test.shape,x_test_poly.shape)

pd.DataFrame(x_train_poly,columns=poly.get_feature_names_out()).head()
```

```python
# 모델 학습
lin_reg = LinearRegression()
lin_reg.fit(x_train_poly,y_train)

# 학습된 회귀 계수, 절편
print(lin_reg.coef_)
print(lin_reg.intercept_)
```

```python
# 모델 평가
# 훈련 데이터는 셋은 잘 맞추지만 평가 데이터 셋에는 맞지 않는 '과대적합' 현상 발생 
print(lin_reg.score(x_train_poly,y_train))
print(lin_reg.score(x_test_poly,y_test))
```
