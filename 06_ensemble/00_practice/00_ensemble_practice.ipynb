```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train_data = pd.read_csv('train.csv')
test_data = pd.read_csv('test.csv')


display(train_data.head())
display(test_data.head())
```

```python
X_train = train_data.iloc[:,:-1]
Y_train = train_data.iloc[:,-1]

X_test = test_data.iloc[:,:-1]
Y_test = test_data.iloc[:,-1]

print(X_train.shape, Y_train.shape)
print(X_test.shape, Y_test.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

Y_train = le.fit_transform(Y_train)
Y_test = le.fit_transform(Y_test)

print(Y_train[:5])
print(Y_test[:5])
```

```python
from sklearn.metrics import  precision_score, recall_score, f1_score, roc_auc_score

def evaluation(model, x, y):
    # 예측값 및 예측 확률 추출
    pred = model.predict(x)
    pred_proba = model.predict_proba(x) # 다중 클래스에서는 전체 확률 사용

    Accuracy = model.score(x, y)
    presicion = precision_score(y, pred, average='macro') 
    recall = recall_score(y, pred, average='macro')
    f1 = f1_score(y, pred, average='macro')
    auc = roc_auc_score(y, pred_proba, multi_class='ovr')

    print(f"Accuracy: {Accuracy:.4f}, Precision: {presicion:.4f}, Recall: {recall:.4f}, F1: {f1:.4f}, AUC: {auc:.4f}")
```

## Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

# 모델 생성 및 학습
lr_clf = LogisticRegression(max_iter=1000)
lr_clf.fit(X_train, Y_train)

# 평가
evaluation(lr_clf, X_test, Y_test)
```

## RandomForest

```python
from sklearn.ensemble import RandomForestClassifier

rf_clf = RandomForestClassifier(random_state=42)
rf_clf.fit(X_train,Y_train)

evaluation(rf_clf,X_test,Y_test)
```

## XGBoost

```python
import xgboost
from xgboost import XGBClassifier

xgb_clf = XGBClassifier()
xgb_clf.fit(X_train, Y_train, eval_set=[(X_train, Y_train), (X_test, Y_test)], verbose=True)

evaluation(xgb_clf,X_test,Y_test)
```

## SVM

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler     
from sklearn.svm import SVC

svm_clf = Pipeline([
    ('scaler', StandardScaler()),
    ('svc', SVC(kernel='linear',probability=True, random_state=42))
])

svm_clf.fit(X_train, Y_train)

evaluation(svm_clf,X_test,Y_test)
```

## KNN

```python
from sklearn.neighbors import KNeighborsClassifier

knn_clf = KNeighborsClassifier()
knn_clf.fit(X_train,Y_train)

evaluation(knn_clf,X_test,Y_test)
```

## RandomizedSearch

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import uniform, randint
from sklearn.decomposition import PCA

# 1. 각 모델별 튜닝할 파라미터 범위(Distributions) 설정
# Pipeline을 사용하는 모델(SVM, KNN)은 '이름__파라미터' 형식을 따릅니다.

param_dicts = [
    {
        'name': 'Logistic Regression',
        'model': lr_clf,
        'params': {
            'C': uniform(0.1, 10),
            'penalty': ['l2'] # solver가 lbfgs일 때 기본
        }
    },
    {
        'name': 'Random Forest',
        'model': rf_clf,
        'params': {
            'n_estimators': randint(50, 200),
            'max_depth': [None, 10, 20, 30],
            'min_samples_split': randint(2, 11)
        }
    },
    {
        'name': 'XGBoost',
        'model': xgb_clf,
        'params': {
            'n_estimators': randint(50, 200),
            'learning_rate': uniform(0.01, 0.3),
            'max_depth': randint(3, 10),
            'gamma': uniform(0, 0.5)
        }
    },
    {
        'name': 'SVM',
        'model': svm_clf, # Pipeline 구조 유지
        'params': {
            'svc__C': uniform(0.1, 10),
            'svc__kernel': ['linear', 'rbf']
        }
    },
    {
        'name': 'KNN',
        'model': Pipeline([('scaler', StandardScaler()), ('knn', KNeighborsClassifier())]),
        'params': {
            'knn__n_neighbors': randint(3, 15),
            'knn__weights': ['uniform', 'distance']
        }
    }
]

# 2. 반복문을 통해 최적화 및 결과 출력
for item in param_dicts:
    print(f"========== {item['name']} Tuning Start ==========")
    
    # n_iter=10: 10개의 조합을 랜덤하게 시도 (데이터 크기에 따라 조절)
    rs = RandomizedSearchCV(
        item['model'], 
        param_distributions=item['params'], 
        n_iter=10, 
        cv=3, 
        scoring='accuracy', 
        n_jobs=-1, 
        random_state=42
    )
    
    rs.fit(X_train, Y_train)
    
    print(f"Best Parameters: {rs.best_params_}")
    
    # 원본 evaluation 함수를 그대로 사용하여 성능 비교
    evaluation(rs.best_estimator_, X_test, Y_test)
    print("\n")
```

## hyperopt

```python
from hyperopt import fmin, tpe, hp, STATUS_OK, Trials
import numpy as np
from sklearn.model_selection import cross_val_score

# 1. 탐색 공간(Space) 정의
# 각 모델별로 파라미터 범위를 설정합니다.
space = hp.choice('classifier_type', [
    {
        'type': 'lr',
        'C': hp.loguniform('lr_C', np.log(0.1), np.log(10.0)),
    },
    {
        'type': 'rf',
        'n_estimators': hp.quniform('rf_n_estimators', 50, 200, 1),
        'max_depth': hp.quniform('rf_max_depth', 5, 30, 1),
        'min_samples_split': hp.quniform('rf_min_samples_split', 2, 10, 1),
    },
    {
        'type': 'xgb',
        'n_estimators': hp.quniform('xgb_n_estimators', 50, 200, 1),
        'max_depth': hp.quniform('xgb_max_depth', 3, 10, 1),
        'learning_rate': hp.uniform('xgb_learning_rate', 0.01, 0.3),
        'gamma': hp.uniform('xgb_gamma', 0, 0.5),
    },
    {
        'type': 'svm',
        'C': hp.loguniform('svm_C', np.log(0.1), np.log(10.0)),
        'kernel': hp.choice('svm_kernel', ['linear', 'rbf']),
    },
    {
        'type': 'knn',
        'n_neighbors': hp.quniform('knn_n_neighbors', 3, 15, 1),
        'weights': hp.choice('knn_weights', ['uniform', 'distance']),
    }
])

# 2. 목적 함수(Objective Function) 정의
def objective(params):
    clf_type = params['type']
    
    if clf_type == 'lr':
        model = LogisticRegression(C=params['C'], max_iter=1000)
    elif clf_type == 'rf':
        model = RandomForestClassifier(
            n_estimators=int(params['n_estimators']),
            max_depth=int(params['max_depth']),
            min_samples_split=int(params['min_samples_split']),
            random_state=42
        )
    elif clf_type == 'xgb':
        model = XGBClassifier(
            n_estimators=int(params['n_estimators']),
            max_depth=int(params['max_depth']),
            learning_rate=params['learning_rate'],
            gamma=params['gamma'],
            random_state=42
        )
    elif clf_type == 'svm':
        model = SVC(C=params['C'], kernel=params['kernel'], probability=True, random_state=42)
    else: # knn
        model = KNeighborsClassifier(
            n_neighbors=int(params['n_neighbors']),
            weights=params['weights']
        )

    # 교차 검증 점수 계산 (정확도)
    score = cross_val_score(model, X_train, Y_train, cv=3, n_jobs=-1).mean()
    
    # Loss는 작을수록 좋으므로 정확도에 -1을 곱함
    return {'loss': -score, 'status': STATUS_OK}

# 3. 최적화 수행
trials = Trials()
best = fmin(fn=objective,
            space=space,
            algo=tpe.suggest,
            max_evals=100, # 총 40회 시도
            trials=trials)

print("\n--- 최적화 완료 ---")
print(best) # 결과값은 인덱스 형태로 출력될 수 있음
```
